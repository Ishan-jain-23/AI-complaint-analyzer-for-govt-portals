```python
import nltk
import re
```

### we will also set up `text` variable here.

```python
text = ""
text
```

### 1. Segmentation

we will break our entirety of text int smalll sentences

```python
#download the initially important package 
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
```

```python
#spliting text into sentences
sentences = sent_tokenize(text)
sentences
```

```python
sentences[0]
```

```python
# REMOVE PUNCTUATION
text = re.sub(r"[^a-zA-Z0-9]"," ", sentences[0])
text
```

## 2. tokenization

simply breaking of our sentences into tokens tokens in this case are words.

```python
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(text)
print(words)
```

## 3 Removal of stop words

`stop words `-> words that appear frequently in launguage, but carry almost none to very minimal meaning for NLP

```python
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
# list of stop words in ENGLISH launguage
print(stopwords.words('english'))
```

```python
sentences[0]
```

```python
# remove stop words from the text
words = [w for w in words if w not in stopwords.words('english')]
print(words)
```

## 4. stemming and lemmatization 
`streming` -> Finding stem root of words 
`Lemetization` -> finding base of the word of the words

```python
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# stemming
from nltk.stem import PorterStemmer


# reduce words to their stem
stemmed = [PorterStemmer().stem(w) for w in words] 
print(stemmed)
```

```python
print(words)
```

```python
#lemmatization
from nltk.stem.wordnet import WordNetLemmatizer

# Reduce words to their base form
lemmatized = [WordNetLemmatizer().lemmatize(w) for w in words]
print(lemmatized)
```

### >>>> Another Helping  example

```python
words2 =['wait','waiting','studies','studying','computer']

#stemming
stemmed2 = [PorterStemmer().stem(w) for w in words2]
print(stemmed2)

#lemmatization
lemmatized2 = [WordNetLemmatizer().lemmatize(w) for w in words2]
print(lemmatized2)  
```

## 5. part of speech tagging 
part of spech tagging is the process of idenfying the part of speech (such as noun , verb etc)


```python
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
```

```python
from nltk import pos_tag

# tag each word with its part of speech
pos_tag(words)
```

NLTK Part of Speech Tag list is as follows:

- CC: It is the conjunction of coordinating
- CD: It is a digit of cardinal
- DT: It is the determiner
- EX: Existential
- FW: It is a foreign word
- IN: Preposition and conjunction
- JJ: Adjective
- JJR and JJS: Adjective and superlative
- LS: List marker
- MD: Modal
- NN: Singular noun
- NNS, NNP, NNPS: Proper and plural noun
- PDT: Predeterminer
- WRB: Adverb of wh
- WP$: Possessive wh
- WP: Pronoun of wh
- WDT: Determiner of wp
- VBZ: Verb
- VBP, VBN, VBG, VBD, VB: Forms of verbs
- UH: Interjection
- TO: To go
- RP: Particle
- RBS, RB, RBR: Adverb
- PRP, PRP$: Pronoun personal and professional

## 6. named entity recoginition (NER)

NER is a means of identifying and categorizing named entities in unstructured text into predefined categories such as names of people,
organizations etc in structured text.

```python
from nltk import ne_chunk
nltk.download('words')
nltk.download("maxent_ne_chunker_tab")
```

```python
ner_tree = ne_chunk(pos_tag(word_tokenize(sentences[0])))  # tokenize the sentence and then chunk it
print(ner_tree)  # print the named entity tree
```
